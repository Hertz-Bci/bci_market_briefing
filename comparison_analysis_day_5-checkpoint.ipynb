```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取两家公司数据
mdt = pd.read_csv('mdt_financials.csv')
jnj = pd.read_csv('jnj_financials.csv')

# 计算研发占比
mdt['rd_ratio'] = mdt['rd_expense'] / mdt['total_revenue']
jnj['rd_ratio'] = jnj['rd_expense'] / jnj['total_revenue']

mdt['year'] = mdt['year'].astype(str)
jnj['year'] = jnj['year'].astype(str)

# 画在同一张图
plt.figure(figsize=(10,5))
plt.plot(mdt['year'],mdt['rd_ratio'],marker='o',linestyle='-',label='Medtronic',color='blue')
plt.plot(jnj['year'],jnj['rd_ratio'],marker='s',linestyle='-',label='J&J',color='orange')

plt.title('R&D Expense Ratio Comparison (2019-2023)')
plt.xlabel('Year')
plt.ylabel('R&D / Revenue')
plt.grid(True)
plt.legend()   # 显示图例，标注哪条线是谁
# Y轴显示百分比，更直观
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y,_:f'{y:.0%}'))
plt.savefig('day5_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```
