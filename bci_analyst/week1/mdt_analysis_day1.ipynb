```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df = pd.read_csv('mdt_financials.csv')
print("数据读取成功，前5行预览")
print(df.head())

# 画折线图
plt.figure(figsize=(8,4))
plt.plot(df['year'],df['rd_expense'],marker='o',linestyle='-')
plt.title('Medtronic R&D Expense (2019-2023)')
plt.xlabel('Year')
plt.ylabel('R&D Expense (100M USD)')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. 解决中文乱码（Windows系统必须加）
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 2. 读取数据
df = pd.read_csv("mdt_financials.csv")

# 3. 新建画布并绘图（这一步必须写在最前面）
plt.figure(figsize=(10, 6))
plt.plot(df['year'], df['rd_expense'], marker='o', linewidth=2, color='#1f77b4')

# 4. 设置图表样式
plt.title('美敦力2019-2023年研发费用趋势', fontsize=14, fontweight='bold')
plt.xlabel('年份', fontsize=12)
plt.ylabel('研发费用（亿美元）', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(df['year'])

# 5. 【关键】先保存图片，再显示！！
plt.savefig('mdt_rd_trend.png', dpi=150, bbox_inches='tight')

# 6. 最后再显示图表
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 解决中文乱码
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 读取数据并处理列名
df = pd.read_csv("mdt_financials.csv")
df.columns = df.columns.str.strip()  # 去除列名前后空格

# 打印列名，方便排查问题
print("列名：", df.columns.tolist())

# 绘图（先保存，再显示）
plt.figure(figsize=(10, 6))
plt.plot(df['year'], df['rd_expense'], marker='o', linewidth=2, color='#1f77b4')

plt.title('美敦力2019-2023年研发费用趋势', fontsize=14, fontweight='bold')
plt.xlabel('年份', fontsize=12)
plt.ylabel('研发费用（亿美元）', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(df['year'])

# 先保存图片
plt.savefig('mdt_rd_trend.png', dpi=150, bbox_inches='tight')
# 再显示图表
plt.show()
```
