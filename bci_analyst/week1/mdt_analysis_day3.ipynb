```python
# Day 3: 美敦力研发投入占比分析
# 目标: 不看绝对金额, 看研发投入占收入的比重

# 1.导入工具库
import pandas as pd                # Pandas:处理表格数据的工具箱
import matplotlib.pyplot as plt      # Matplotlib:画图的工具箱

# 2.读取数据
# pd.read_csv() 是"读取CSV文件的咒语" 
# df 是 DateFrame的缩写,代表"一张表格"
df = pd.read_csv('mdt_financials.csv')

# 3.看一眼数据,确保读对了
print("数据预览:")
print(df)
print("\n")   # 打印一个空行，让输出更清晰

# 4. 计算新列:研发占比
# df['rd_radio'] 的意思是: "在表格里新建一列,名字叫 rd_radio"
df['rd_radio'] = df['rd_expense'] / df['total_revenue']
print("研发占比计算结果:")
print(df[['year','total_revenue','rd_expense','rd_radio']])

# 5. 画出研发占比的折线图
plt.figure(figsize=(8,4))   #设置画布大小,宽8英寸,高4英寸
plt.plot(df['year'],df['rd_ratio'],marker='s',linestyle='-',color='green')

# 6设置图表标题和轴标签
plt.title('Medtronic R&D Expense Ratio (2019-2023)')
plt.xiabel('Year')
plt.ylabel('R&D / Revenue')
plt.grid(True)
plt.show
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. 读取数据
df = pd.read_csv('mdt_financials.csv')
print("数据预览：")
print(df)

# 2. 计算新列：研发占比 = 研发费用 / 总收入
df['rd_ratio'] = df['rd_expense'] / df['total_revenue']
print("\n研发占比计算结果：")
print(df[['year', 'total_revenue', 'rd_expense', 'rd_ratio']])

# 3. 画出研发占比的折线图
plt.figure(figsize=(8, 4))
plt.plot(df['year'], df['rd_ratio'], marker='s', linestyle='-', color='green')
plt.title('Medtronic R&D Expense Ratio (2019-2023)')
plt.xlabel('Year')
plt.ylabel('R&D / Revenue')
plt.grid(True)
plt.show()
```

1. 2019-2021年美敦力研发绝对金额与占比同向上升，2021年后金额微升但占比回落，本质是前者反映投入规模，后者反映投入的相对强度。
2. 研发绝对金额下降但占比稳定或上升，说明公司营收下滑幅度大于研发降幅，是业务承压下守住研发战略底线的被动表现。
3. 仍坚持长期加大研发投入的大方向，但建议以8.3%为研发占比底线，根据营收增速动态调整投入节奏，而非追求绝对金额刚性增长。
