```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. 读取数据
df = pd.read_csv('jnj_financials.csv')
print("数据预览：")
print(df)

# 2. 计算新列：研发占比 = 研发费用 / 总收入
df['rd_ratio'] = df['rd_expense'] / df['total_revenue']
print("\n研发占比计算结果：")
print(df[['year', 'total_revenue', 'rd_expense', 'rd_ratio']])

# 3. 画出研发占比的折线图
plt.figure(figsize=(8, 4))
plt.plot(df['year'], df['rd_ratio'], marker='s', linestyle='-', color='orange')
plt.title('JNJ R&D Expense Ratio (2019-2023)')
plt.xlabel('Year')
plt.ylabel('R&D / Revenue')
plt.grid(True)
plt.show()
```

强生在2021年战略性提升研发和投入强度、后续年份虽略有收缩，但研发费用占营收比例长期高于2019-2020基准，整体研发投入力度相较周期前期明显增强。与美敦力相比，强生在2021年选择逆势加码研发，这种“在营收承压时强化长期投入”的策略，在BCI等新兴赛道竞争加剧的背景下，可能为其赢得更有力的技术卡位。
